```python
# Using a list for storing account information:

accountNamesList = []
accountBalancesList = []
accountPasswordsList = []

def newAccount(name, balance, password):
    global accountNamesList, accountBalancesList, accountPasswordsList
    accountNamesList.append(name)
    accountBalancesList.append(balance)
    accountPasswordsList.append(password)

def show(accountNumber):
    global accountNamesList, accountBalanceList, accountPasswordsList
    print(
        f'{'='*40}\n'
        f'Account, {accountNumber}\n'
        f'       Name: {accountNamesList[accountNumber]:>11}\n'
        f'       Balance: {accountBalancesList[accountNumber]:>6}\n'
        f'       Password: {accountPasswordsList[accountNumber]:>5}\n'
        f'{'='*40}\n'
    )

def getBalance(accountNumber, password):
    global accountNamesList, accountBalancesList, accountPasswordsList
    if password != accountPasswordsList[accountNumber]:
        print('Incorrect password')
        return None
    return accountBalancesList[accountNumber]

def deposit(accountNumber, amountToDeposit, password):
    global accountNamesList, accountBalancesList, accountPasswordsList
    if amountToDeposit[accountNumber] < 0:
        print('You cannot deposit a negative amount!')
        return None

    if password != accountPasswordsList[accountNumber]:
        print('Incorrect password')
        return None

    accountBalance[accountNumber] = accountBalance[accountNumber] + amountToDeposit[accountNumber]
    return accountBalance[accountNumber]

def withdraw(accountNumber, amountToWithdraw, password):
    global accountNamesList, accountBalancesList, accountPasswordsList
    if amountToWithdraw[accountNumber] < 0:
        print('You cannot withdraw a negative amount')
        return None

    if password != accountPasswordsList[accountNumber]:
        print('Incorrect password for this account')
        return None

    if amountToWithdraw[accountNumber] > accountBalance[accountNumber]:
        print('You cannot withdraw more than you have in your account')
        return None

    accountBalance[accountNumber] = accountBalance[accountNumber] - amountToWithdraw[accountNumber]
    return accountBalance[accountNumber]

# Add accounts
print(f"Your account number is: {len(accountNamesList)}")
newAccount('Francis', 1000, 'science')


print(f"Your account number is: {len(accountNamesList)}")
newAccount('Clyde', 500, 'marketing')

# Print all accounts (test)
for i in range(len(accountNamesList)):
    show(i)


# Bank run time:
while True:
    print(
        f"\n"
        f"Press b to get the balance\n"
        f"Press d to make a deposit\n"
        f"Press w to make a withdrawal\n"
        f"Press s to show the account\n"
        f"Press q to quit\n"
        f"\n"
    )

    action = input('What do you want to do? ')
    action = action.lower()
    action = action[0]
    print()

    if action == 'b':
        print('Get Balance:')
        userAccountNumber = input('Please enter your account number: ')
        userAccountNumber = int(userAccountNumber)
        userPassword = input('Please enter the password: ')
        theBalance = getBalance(userAccountNumber,userPassword)
        if theBalance is not None:
            print(f'Your balance is: {theBalance}')

    elif action == 'd':
        print('Deposit:')
        userAccountNumber = input('Please enter your account number: ')
        userAccountNumber = int(userAccountNumber)
        userDepositAmount = input('Please enter amount to deposit: ')
        userDepositAmount = int(userDepositAmount)
        userPassword = input('Please enter the password: ')

        newBalance = deposit(userAccountNumber, userDepositAmount, userPassword)
        if newBalance is not None:
            print(f'Your new balance is: {newBalance}')

    elif action == 'w':
        print('Withdraw:')
        userAccountNumber = input('Please enter your account number: ')
        userAccountNumber = int(userAccountNumber)
        userWithdrawAmount = input('Please enter amount to withdraw: ')
        userWithdrawAmount = int(userWithdrawAmount)
        userPassword = input('Please enter the password: ')

        remainingBalance = withdraw(userAccountNumber,userWithdrawAmount, userPassword)
        if remainingBalance is not None:
            print(f'Your remaining balance is: {remainingBalance}')


    elif action == 's':
        userAccountNumber = input('Please enter your account number: ')
        userAccountNumber = int(userAccountNumber)
        userPassword = input('Please enter the password: ')
        if userPassword == accountPassword:
            print(f'{'='*10}')
            show(userAccountNumber)
            print(
                f'Show current account details:\n'
                f'Here is your current balance: {accountBalance}'
            )
            print(f'{'='*10}')

        else:
            print('Incorrect password, quitting...')
            break

    elif action == 'q':
        print('Exiting...')
        break

    ### For test and development
    elif action == 'p':
        print(
            f'Account Names List: {accountNamesList}\n'
            f'Account Names List: {accountBalancesList}\n'
            f'Account Names List: {accountPasswordsList}\n'
        )
        break


print('Done')
```
