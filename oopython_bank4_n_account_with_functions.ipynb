```python
accountNamesList = []
accountBalancesList = []
accountPasswordsList = []

def newAccount(name, balance, password):
    global accountNamesList, accountBalancesList, accountPasswordsList
    accountNamesList.append(name)
    accountBalancesList.append(balance)
    accountPasswordsList.append(password)

def show(accountNumber):
    global accountNamesList, accountBalanceList, accountPasswordsList
    print(
        f'{'='*40}\n'
        f'Account, {accountNumber}\n'
        f'       Name: {accountNamesList[accountNumber]:>11}\n'
        f'       Balance: {accountBalancesList[accountNumber]:>6}\n'
        f'       Password: {accountPasswordsList[accountNumber]:>5}\n'
        f'{'='*40}\n'
    )

newAccount('Francis', 100, 'xxx')
newAccount('Clyde', 100, 'xxx')
show(1)
```
